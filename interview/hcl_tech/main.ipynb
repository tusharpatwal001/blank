```python
#  Find the first non-repeating character in a string without using count(), Counter, dictionaries, or set()
S = 'aabbcdde'

def first_non_repeating(s: str):
    first_non_repeating_ = None

    n = len(s)
    for i in range(n):
        occurance = 0
        for j in range(n):
            if s[j]==s[i]:
                occurance +=1
        if occurance == 1:
            first_non_repeating_= s[i]
            break
    return first_non_repeating_


first_non_repeating(S)
```

```python
def func(s: str):
    n = len(s)
    for i in range(n):
        occurance = 0
        for j in range(n):
            if s[j] == s[i]:
                occurance+=1
        if occurance==1:
            return s[i]

S = "aabbccdd"
func(S)
```

```python
# Count the number of employees under each manager whose salary > 500. 
employees = [
    {"id": 1, "name": "Alice", "salary": 600, "manager_id": 10},
    {"id": 2, "name": "Bob", "salary": 400, "manager_id": 10},
    {"id": 3, "name": "Charlie", "salary": 700, "manager_id": 11},
    {"id": 4, "name": "David", "salary": 550, "manager_id": 10},
    {"id": 5, "name": "Eve", "salary": 450, "manager_id": 11},
    {"id": 6, "name": "Frank", "salary": 800, "manager_id": 12},
    {"id": 7, "name": "Grace", "salary": 500, "manager_id": 10},
    {"id": 8, "name": "Heidi", "salary": 900, "manager_id": 12},
]

def count_employees(employees):
    result = {}

    for emp in employees:
        if emp["salary"] > 500:
            manager = emp["manager_id"]
            result[manager] = result.get(manager, 0) + 1

    return result

print(count_employees(employees))
```

```python
def emp_calary_count(employees: list):

    managers = {}
    for i in employees:
        if i['salary'] > 500:
            manager = i['manager_id']
            if manager in managers:
                managers[manager] += 1
            else:
                managers[manager] = 1
    return managers

emp_calary_count(employees)    
```
