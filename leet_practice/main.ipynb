```python
# 1863. Sum of All Subset XOR Totals
def subsetXORSum(nums: list[int]):
    subs = [[]]
    for ele in nums:
        subs += [i + [ele] for i in subs]

    res = 0
    for ele in subs:
        if len(ele)==1:
            res+=ele[0]
        elif len(ele)>1:
            res += eval(" ^ ".join([str(i) for i in ele]))

    return res

subsetXORSum([5, 1, 6])
        
```

```python
# 3658. GCD of Odd and Even Sums
def gcdOfOddEvenSums(n: int):
    sumodd = []
    sumeven = []

    sample_n = n
    i = 1
    while sample_n > 0:

        if i % 2 != 0:
            sumodd.append(i)
            sample_n -= 1
        i += 1

    sample_n = n
    i = 1
    while sample_n > 0:
        if i % 2 == 0:
            sumeven.append(i)
            sample_n -= 1
        i += 1

    return abs(sum(sumodd) - sum(sumeven))


gcdOfOddEvenSums(4)
```

```python
# 3701. Compute Alternating Sum
def alternatingSum(nums: list[int]) -> int:
        res = 0
        n = len(nums)
        for i in range(n):
            if i % 2 != 0:
                res -= nums[i]
            else:
                res += nums[i]
        return res

alternatingSum([1,3,5,7])
```

```python
# 1470. Shuffle the Array
def shuffle(nums: list[int], n: int):
    res = [item for pair in zip(nums[:n], nums[n:]) for item in pair]
    return res


shuffle([2, 5, 1, 3, 4, 7], n=3)
```

```python
# 1684. Count the Number of Consistent Strings
def countConsistentStrings(allowed: str, words: list[str]):
    res = 0
    for word in words:
        if len([1 for char in word if char in allowed]) == len(word):
            res+=1
    return res



countConsistentStrings(allowed = "ab", words = ["ad","bd","aaab","baa","badab"])
```

```python
# 2824. Count Pairs Whose Sum is Less than Target
def countPairs(nums: list[int], target: int) -> int:
    n = len(nums)
    res = 0
    for i in range(n):
        for j in range(n):
            if i != j and nums[i] + nums[j] < target:
                print(f"{nums[i]=} + {nums[j]=} < {target=}" )
                res += 1
    return res

countPairs(nums = [-1,1,2,3,1], target = 2)
```

```python
class Solution:
    def numberGame(self, nums: list[int]):
        alice = []
        bob = []
        new = nums
        for i in range(len(nums)):
            if len(alice) <= len(bob):
                val = min(new)
                alice.append(val)
                new.remove(val)
            else:
                val = min(new)
                bob.append(val)
                new.remove(val)

        res = []
        for i, j in zip(bob, alice):
            res.append(i)
            res.append(j)

        return res

v = Solution()
v.numberGame([5,4,2,3])
```

```python
class Solution:
    def numberOfMatches(self, n: int):
        res = 0
        while n > 1:
            if n % 2 == 0:
                res += n // 2
                n = n // 2
            else:
                res += (n - 1) // 2
                n = (n - 1) // 2 + 1
        return res


s = Solution()
s.numberOfMatches(7)
```

```python
def divide(dividend: int, divisor: int):
    print(f"{dividend=} // {divisor=} = {dividend//divisor=}")
    sign = 1
    if dividend < 0:
        sign = sign * -1
        dividend = dividend * sign
    elif divisor < 0:
        sign = sign * -1
        divisor = divisor * sign

    return sign * (dividend // divisor)


divide(7, -3)
```
